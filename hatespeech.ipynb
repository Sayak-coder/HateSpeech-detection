```python
import pandas as pd
import numpy as np
import transformers
from transformers import pipeline
```

```python
df = pd.read_csv('/content/drive/MyDrive/StereoQueerEval/Dataset from IIT Hydrabad/Wiki Madlibs” Eval Dataset/hi-Latn_sentence_templates.csv')
df.head()
```

```python
df['toxicity'].unique()
```

```python
df['template'].unique()
```

```python
df['toxicity'] = df['toxicity'].map({'nontoxic': 0, 'toxic': 1})
df.head()
```

```python
df['template'].values
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tv = TfidfVectorizer(max_features=2500)
```

```python
from sklearn.model_selection import GroupKFold

X = tv.fit_transform(df['phrase']).toarray()
y=df['toxicity']
groups = df['template'].values

gkf=GroupKFold(n_splits=4)

fold = 1
for train_idx, test_idx in gkf.split(X, y, groups=groups):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
```

```python
from sklearn.naive_bayes import MultinomialNB
hatespeech_detect_model = MultinomialNB().fit(X_train, y_train)
```

```python
y_pred=hatespeech_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,y_pred)
print(score)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```
