```python
import pandas as pd
import numpy as np
import transformers
from transformers import pipeline
```

```python
df = pd.read_csv('/content/drive/MyDrive/StereoQueerEval/Dataset from IIT Hydrabad/Wiki Madlibs” Eval Dataset/hi-Latn_sentence_templates.csv')
df.head()
```

```python
df['toxicity'].unique()
```

```python
df = df.drop(columns='template')
df.columns
```

```python
df.shape
```

```python
df['toxicity'] = df['toxicity'].map({'nontoxic': 0, 'toxic': 1})
df.head()
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
nltk.download('punkt_tab')
```

```python
sentences=[]
for sentence in df['phrase'] :
  sentences.append(sentence)
print(len(sentences))
```

```python
stemmer=PorterStemmer()
```

```python
import re
corpus=[]
for i in range(len(sentences)):
  review = re.sub('[^a-zA-Z]',' ',sentences[i])
  review = review.lower()
  corpus.append(review)
```

```python
nltk.download('stopwords')
```

```python
for i in corpus:
  words = nltk.word_tokenize(i)
  for word in words:
    print(stemmer.stem(word))
```

```python
y=df['toxicity']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=2500)
X = cv.fit_transform(corpus).toarray()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```python
from sklearn.naive_bayes import MultinomialNB
hatespeech_detect_model = MultinomialNB().fit(X_train, y_train)
```

```python
y_pred=hatespeech_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,y_pred)
print(score)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```

```python
from xgboost import XGBClassifier
classifier = XGBClassifier()
classifier.fit(X_train, y_train)
y_pred=classifier.predict(X_test)
print(classification_report(y_pred,y_test))
```
