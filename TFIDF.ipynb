```python
import pandas as pd
```

```python
df= pd.read_csv('/content/StereoQueerEval_EN_training.tsv', sep='\t')
df
```

```python
df.isnull().sum()
```

```python
df=df.drop(['StereoQueerEval_id','hate_speech','target'],axis=1)
df.columns
```

```python
X=df.drop('stereotype',axis=1)
y=df['stereotype']
```

```python
X.head()
```

```python
y.head()
```

```python
y=y.map({'no':0,'yes':1})
y.unique()
```

```python
import re
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
corpus_title=[]
corpus_description=[]
corpus_comment=[]

stemmer=PorterStemmer()

for i in range(0,len(df['yt_title'])):
  review=re.sub('[^a-zA-Z]',' ',df['yt_title'][i])
  review=review.lower()
  review=review.split()

  review=(stemmer.stem(word) for word in review if not word in stopwords.words('english'))
  review=' '.join(review)
  corpus_title.append(review)

for i in range(0,len(df['yt_description'])):
  review=re.sub('[^a-zA-Z]',' ',df['yt_description'][i])
  review=review.lower()
  review=review.split()

  review=(stemmer.stem(word) for word in review if not word in stopwords.words('english'))
  review=' '.join(review)
  corpus_description.append(review)

for i in range(0,len(df['yt_comment'])):
  review=re.sub('[^a-zA-Z]',' ',df['yt_comment'][i])
  review=review.lower()
  review=review.split()

  review=(stemmer.stem(word) for word in review if not word in stopwords.words('english'))
  review=' '.join(review)
  corpus_comment.append(review)
```

```python
print(X.shape)
print(len(corpus_title))
print(len(corpus_description))
print(len(corpus_comment))
```

```python
tv = TfidfVectorizer()
corpus_title_TFIDF=tv.fit_transform(corpus_title)
corpus_description_TFIDF=tv.fit_transform(corpus_description)
corpus_comment_TFIDF=tv.fit_transform(corpus_comment)
```

```python
print(corpus_title_TFIDF.toarray())
print(corpus_description_TFIDF.toarray())
print(corpus_comment_TFIDF.toarray())
```

```python
import numpy as np

# Convert each dense TF-IDF array row into a list to store it in a single column cell
X['yt_title'] = list(corpus_title_TFIDF.toarray())
X['yt_description'] = list(corpus_description_TFIDF.toarray())
X['yt_comment'] = list(corpus_comment_TFIDF.toarray())

print("X shape:", X.shape)
display(X.head())
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```python
import numpy as np
from sklearn.naive_bayes import MultinomialNB

X_train_title = np.stack(X_train['yt_title'].values)
X_train_desc = np.stack(X_train['yt_description'].values)
X_train_comm = np.stack(X_train['yt_comment'].values)

X_train = np.hstack([X_train_title, X_train_desc, X_train_comm])

model = MultinomialNB().fit(X_train, y_train)
```

```python
X_test_title = np.stack(X_test['yt_title'].values)
X_test_desc = np.stack(X_test['yt_description'].values)
X_test_comm = np.stack(X_test['yt_comment'].values)

X_test = np.hstack([X_test_title, X_test_desc, X_test_comm])
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
